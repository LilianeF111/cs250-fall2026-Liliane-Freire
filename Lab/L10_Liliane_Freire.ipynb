```python
#Text Validation - Find the words that mirror each other

words = ["stop", "pots", "live", "evil",
"python", "level", "dog"]

# 1. Reverse every word
backwards = [word[::-1] for word in words]

# 2. Keep words whose reverse is also in words
mirrored = [word for word in words if word[::-1] in words]

# 3. Separate palindromes from true pairs
pals = [word for word in mirrored if word == word[::-1]]

pairs = [word for word in mirrored if word != word[::-1]]

print("mirrored ->", mirrored)
print("pals ->", pals)
print("pairs ->", pairs)
```

```python
#Palindrome in list
words = ["Racecar", "python", "Level", "noon", "hello"]

palindromes = [word for word in words
               if word.lower() == word.lower()[::-1]]

print(palindromes)
```

```python
#numerical palindorme
number = 1221

working_number = number
reversed_number = 0

while working_number > 0:
    digit = working_number % 10
    reversed_number = reversed_number * 10 + digit
    working_number //= 10

is_palindrome = number == reversed_number

print(is_palindrome)
```

```python
#Palindrome phrase
phrase = "Was it a car or a cat I saw"

cleaned = ""

for ch in phrase:
    if ch.isalnum():
        cleaned += ch.lower()

is_palindrome = cleaned == cleaned[::-1]

print(is_palindrome)
```

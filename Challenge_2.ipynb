```python
#Challenge_2

def topKFrequent(nums, k):
    frequency = {}

    # Step 1: Count each number
    for num in nums:
        if num in frequency:
            frequency[num] += 1
        else:
            frequency[num] = 1

    # Step 2: Sort by frequency
    sorted_nums = sorted(
        frequency,
        key=frequency.get,
        reverse=True
    )

    # Step 3: Return the top k numbers
    return sorted_nums[:k]


nums = [1, 1, 1, 2, 2, 3]
k = 2

print(topKFrequent(nums, k))
```

```python
#Code without sorted() or sort()

#max(frequency, key=frequency.get) -> finds the key with the highest frequency.
#result.append(most_frequent) -> adds the found number to the list of results
#del frequency[most_frequent] -> removes that number from the dictionary so it is not selected again.

def topKFrequent(nums, k):
    frequency = {}

    for num in nums:
        if num in frequency:
            frequency[num] += 1
        else:
            frequency[num] = 1

    result = []

    for i in range(k):
        most_frequent = max(frequency, key=frequency.get)

        result.append(most_frequent)

        del frequency[most_frequent]

    return result


nums = [1, 1, 1, 2, 2, 3]
k = 2

print(topKFrequent(nums, k))
```

# VeteransNet - Final Python Project (CS250)
VeteransNet is a simple Python program designed to help U.S. veterans find useful resources in different categories, including mental health, education, sports, community activities, and benefits.

How to run: Click Runtime > Run all in Google Colab. Select a category from the menu (1–5) or enter 0 to exit. A graph showing the available resources will appear at the end.

## 1. Main Program Functions

```python
def display_menu(categories):
    """Print the categories and exit option."""
    print("\n=== VeteransNet Resource Finder ===")
    for number, category in enumerate(categories, start=1):
        print(f"{number}. {category}")
    print("0. Exit")


def search_resources(category, resources):
    """Return resources in the selected category."""
    return resources.get(category, [])


def display_results(results):
    """Show matching resources or a no-results message."""
    if not results:
        print("No resources found in this category.")
    else:
        print("\nAvailable resources:")
        for name, website in results:
            print(f"- {name}: {website}")


def main():
    """Create the directory and repeat the menu until Exit."""
    resources = {'Mental Health & Wellness': [('VA Mental Health', 'https://www.mentalhealth.va.gov/'), ('Veterans Crisis Line', 'https://www.veteranscrisisline.net/')], 'Community & Networking': [('Team RWB', 'https://teamrwb.org/'), ('The Mission Continues', 'https://www.missioncontinues.org/')], 'Sports & Activities': [('Move United', 'https://moveunitedsport.org/'), ('Team RWB Activities', 'https://teamrwb.org/')], 'Education & Career': [('VA Education Benefits', 'https://www.va.gov/education/'), ('VA Careers and Employment', 'https://www.va.gov/careers-employment/')], 'Benefits & Discounts': [('VA Benefits', 'https://www.va.gov/'), ('Veterans Canteen Service', 'https://www.vacanteen.va.gov/')]}
    categories = list(resources.keys())
    while True:
        display_menu(categories)
        choice = input("Choose a number (0-5): ").strip()
        if choice == "0":
            print("Thank you for using VeteransNet!")
            break
        elif choice.isdigit() and 1 <= int(choice) <= len(categories):
            category = categories[int(choice) - 1]
            print(f"\nSelected: {category}")
            results = search_resources(category, resources)
            display_results(results)
        else:
            print("Invalid selection. Please enter a number from 0 to 5.")
    show_results_graph(resources)
```

## 2. Results graph

```python
# Define the graph function. It receives the directory from main().
import matplotlib.pyplot as plt


def show_results_graph(resources):
    categories = list(resources.keys())
    counts = [len(resources[category]) for category in categories]
    plt.figure(figsize=(9, 4))
    plt.barh(categories, counts)
    plt.xlabel("Number of sample resources")
    plt.title("VeteransNet: Resources by Category")
    plt.xlim(0, max(counts) + 1)
    plt.tight_layout()
    plt.show()
    print("Total resource entries:", sum(counts))
```

## 3. Live demonstration



```python
# Run this cell to start the interactive menu.
main()
```
